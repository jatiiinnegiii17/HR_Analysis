```python
import pandas as pd
df=pd.read_csv("hr_raw_data.csv")
df.head()
```

```python
print("Total rows:", len(df))
```

```python
import matplotlib.pyplot as plt

```

```python
department_count=df["department"].value_counts()
plt.figure(figsize=(10,6))
plt.bar(
    department_count.index,
    department_count.values
)
plt.title("Employees Headcount by department")
plt.xlabel("Department")
plt.ylabel("Headcount")
plt.xticks(rotation=45)
plt.show()
```

```python
job_level_count = df["job_level"].value_counts()

plt.figure(figsize=(8, 5))

plt.bar(
    job_level_count.index,
    job_level_count.values
)

plt.title("Employee Headcount by Job Level")
plt.xlabel("Job Level")
plt.ylabel("Headcount")

plt.tight_layout()
plt.show()
```

```python

status_count = df["status"].value_counts()

plt.figure(figsize=(8, 5))

plt.bar(
    status_count.index,
    status_count.values
)

plt.title("Employee Status Distribution")
plt.xlabel("Status")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()
```

```python
work_mode_count = df["work_mode"].value_counts()

plt.figure(figsize=(8, 5))

plt.bar(
    work_mode_count.index,
    work_mode_count.values
)

plt.title("Work Mode Distribution")
plt.xlabel("Work Mode")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()
```

```python
avg_salary_department = (
    df.groupby("department")["salary"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

plt.bar(
    avg_salary_department.index,
    avg_salary_department.values
)

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
avg_salary_level = (
    df.groupby("job_level")["salary"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

plt.bar(
    avg_salary_level.index,
    avg_salary_level.values
)

plt.title("Average Salary by Job Level")
plt.xlabel("Job Level")
plt.ylabel("Average Salary")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df["age"],
    bins=20
)

plt.title("Employee Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df["salary"],
    bins=50
)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()
```

```python
performance_count = df["performance_rating"].value_counts()

plt.figure(figsize=(10, 6))

plt.bar(
    performance_count.index,
    performance_count.values
)

plt.title("Performance Rating Distribution")
plt.xlabel("Performance Rating")
plt.ylabel("Number of Employees")

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
attrition = (
    df.groupby("department")["status"]
    .apply(lambda x: (x == "Resigned").mean() * 100)
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

plt.bar(
    attrition.index,
    attrition.values
)

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
attrition_level = (
    df.groupby("job_level")["status"]
    .apply(lambda x: (x == "Resigned").mean() * 100)
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

plt.bar(
    attrition_level.index,
    attrition_level.values
)

plt.title("Attrition Rate by Job Level")
plt.xlabel("Job Level")
plt.ylabel("Attrition Rate (%)")

plt.tight_layout()
plt.show()
```

```python
attrition_work_mode = (
    df.groupby("work_mode")["status"]
    .apply(lambda x: (x == "Resigned").mean() * 100)
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

plt.bar(
    attrition_work_mode.index,
    attrition_work_mode.values
)

plt.title("Attrition Rate by Work Mode")
plt.xlabel("Work Mode")
plt.ylabel("Attrition Rate (%)")

plt.tight_layout()
plt.show()
```

```python
hiring_trend = (
    df.groupby("hire_year")
    .size()
)

plt.figure(figsize=(10, 6))

plt.plot(
    hiring_trend.index,
    hiring_trend.values,
    marker="o"
)

plt.title("Employee Hiring Trend")
plt.xlabel("Hire Year")
plt.ylabel("Number of Hires")

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
sample_df = df.sample(
    n=10000,
    random_state=42
)

plt.figure(figsize=(10, 6))

plt.scatter(
    sample_df["experience_years"],
    sample_df["salary"],
    alpha=0.3
)

plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")

plt.tight_layout()
plt.show()
```

```python
performance_salary = (
    df.groupby("performance_rating")["salary"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

plt.bar(
    performance_salary.index,
    performance_salary.values
)

plt.title("Average Salary by Performance Rating")
plt.xlabel("Performance Rating")
plt.ylabel("Average Salary")

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
